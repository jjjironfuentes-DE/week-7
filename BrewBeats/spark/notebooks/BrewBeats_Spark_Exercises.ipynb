## question_1_rdd


```python
import csv
from pyspark.sql import SparkSession

# 1. Create Spark session
spark = SparkSession.builder \
    .appName("BrewBeats_Question_1") \
    .master("local[*]") \
    .getOrCreate()

# 2. Read Spotify CSV file as an RDD
spotify_rdd = spark.sparkContext.textFile(
    "data/raw/Spotify_data.csv"
)

# 3. Remove the header
header = spotify_rdd.first()

spotify_data = spotify_rdd.filter(
    lambda row: row != header
)

# 4. Read CSV rows correctly
rows = spotify_data.map(
    lambda row: next(csv.reader([row]))
)

# 5. Extract artist and count songs
artist_counts = rows.map(
    lambda row: (row[16].strip(), 1)
).reduceByKey(
    lambda a, b: a + b
)

# 6. Sort artists by number of songs
top_artists = artist_counts.sortBy(
    lambda item: item[1],
    ascending=False
).take(20)

# 7. Show top 20 artists
print("\nTOP 20 ARTISTS BY NUMBER OF SONGS")

for artist, count in top_artists:
    print(artist, count)

spark.stop()
```

## question_2_rdd


```python
from pyspark.sql import SparkSession
import csv

# 1. Create Spark session
spark = SparkSession.builder \
    .appName("BrewBeats_Question_2") \
    .master("local[*]") \
    .getOrCreate()

# 2. Read Spotify CSV file
spotify_rdd = spark.sparkContext.textFile(
    "data/raw/Spotify_data.csv"
)

# 3. Remove header
header = spotify_rdd.first()

spotify_data = spotify_rdd.filter(
    lambda row: row != header
)

# 4. Parse CSV rows correctly
rows = spotify_data.map(
    lambda row: next(csv.reader([row]))
)

# 5. Extract danceability values
danceability = rows.map(
    lambda row: float(row[2])
)

# 6. Calculate sum and count
total = danceability.sum()
count = danceability.count()

# 7. Calculate average
average = total / count

# 8. Print result
print("\nDANCEABILITY ANALYSIS")
print("Total songs:", count)
print("Average danceability:", round(average, 4))

spark.stop()
```

## question_3_dataframe


```python
from pyspark.sql import SparkSession

# 1. Create Spark session
spark = SparkSession.builder \
    .appName("BrewBeats_Question_3") \
    .master("local[*]") \
    .getOrCreate()

# 2. Read coffee sales CSV
coffee_df = spark.read.csv(
    "data/raw/coffee_sales.csv",
    header=True,
    inferSchema=True
)

# 3. Show the first five rows
coffee_df.show(5)

# 4. Show the DataFrame structure
coffee_df.printSchema()

# 5. Calculate total revenue
from pyspark.sql.functions import sum

total_revenue = coffee_df.agg(
    sum("money").alias("total_revenue")
)

# 6. Show the result
print("\nTOTAL COFFEE REVENUE")

total_revenue.show()
```

## question_4_dataframe


```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import count

# 1. Create Spark session
spark = SparkSession.builder \
    .appName("BrewBeats_Question_4") \
    .master("local[*]") \
    .getOrCreate()

# 2. Read coffee sales CSV
coffee_df = spark.read.csv(
    "data/raw/coffee_sales.csv",
    header=True,
    inferSchema=True
)

# 3. Count transactions by coffee name
coffee_counts = coffee_df.groupBy(
    "coffee_name"
).agg(
    count("*").alias("total_transactions")
)

# 4. Sort results
coffee_counts = coffee_counts.orderBy(
    "total_transactions",
    ascending=False
)

# 5. Show results
print("\nTRANSACTIONS BY COFFEE")

coffee_counts.show()

spark.stop()
```

## question_5_6_dataframe


```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum, date_format

# 1. Create Spark session
spark = SparkSession.builder \
    .appName("BrewBeats_Question_5") \
    .master("local[*]") \
    .getOrCreate()

# 2. Read coffee sales CSV
coffee_df = spark.read.csv(
    "data/raw/coffee_sales.csv",
    header=True,
    inferSchema=True
)

# 3. Extract month from date
coffee_df = coffee_df.withColumn(
    "month",
    date_format("date", "yyyy-MM")
)

# 4. Calculate monthly revenue
monthly_revenue = coffee_df.groupBy(
    "month"
).agg(
    sum("money").alias("monthly_revenue")
)

# 5. Sort by month
monthly_revenue = monthly_revenue.orderBy("month")

# 6. Show results
print("\nMONTHLY REVENUE")

monthly_revenue.show()

spark.stop()

##### RDD vs DataFrame

#RDD (Resilient Distributed Dataset) is a low-level Spark structure that allows developers to process distributed data using transformations and actions.

#DataFrame is a higher-level structure that organizes data into named columns, similar to a database table.
#RDD provides more control over data processing, while DataFrames offer better optimization and easier data manipulation.
#DataFrames are generally preferred for structured data analysis.
```

## question_7_rdd


```python
from pyspark.sql import SparkSession
import csv

# 1. Create Spark session
spark = SparkSession.builder \
    .appName("BrewBeats_Question_7") \
    .master("local[*]") \
    .getOrCreate()

# 2. Read Spotify CSV file
spotify_rdd = spark.sparkContext.textFile(
    "data/raw/Spotify_data.csv"
)

# 3. Remove header
header = spotify_rdd.first()

spotify_data = spotify_rdd.filter(
    lambda row: row != header
)

# 4. Parse CSV rows correctly
rows = spotify_data.map(
    lambda row: next(csv.reader([row]))
)

# 5. Extract song titles
song_titles = rows.map(
    lambda row: row[15]
)

# 6. Filter titles containing "love"
love_songs = song_titles.filter(
    lambda title: "love" in title.lower()
)

# 7. Count matching songs
total_love_songs = love_songs.count()

# 8. Print result
print("\nSONG TITLES CONTAINING LOVE")
print("Total songs containing 'love':", total_love_songs)

spark.stop()
```

## question_8_playlist


```python
from pyspark.sql import SparkSession
import csv

# 1. Create Spark session
spark = SparkSession.builder \
    .appName("BrewBeats_Question_8") \
    .master("local[*]") \
    .getOrCreate()

# 2. Read Spotify CSV
spotify_rdd = spark.sparkContext.textFile(
    "data/raw/Spotify_data.csv"
)

# 3. Remove header
header = spotify_rdd.first()

spotify_data = spotify_rdd.filter(
    lambda row: row != header
)

# 4. Parse CSV correctly
rows = spotify_data.map(
    lambda row: next(csv.reader([row]))
)

# 5. Extract relevant information
songs = rows.map(
    lambda row: (
        row[15],          # Song title
        row[16],          # Artist
        float(row[2]),    # Danceability
        float(row[3]),    # Duration
        float(row[4])     # Energy
    )
)

# 6. Songs selected for the children's birthday playlist
approved_songs = [
    ("The Happy Song", "Jamie Grace"),
    ("My Favorite Snack", "Imagination Movers"),
    ("When Can I See You Again?", "Owl City"),
    ("Try Everything", "Shakira"),
    ("I2I", "Tevin Campbell"),
    ("Cha Cha Slide", "Mr. C"),
    ("Agadoo", "Black Lace"),
    ("Das Lied Der Schluempfe", "Vader Abraham")
]

# 7. Filter only approved songs
birthday_playlist = songs.filter(
    lambda song: (song[0], song[1]) in approved_songs
)

# 8. Display final playlist
print("\nCHILDREN'S BIRTHDAY PLAYLIST")

for song in birthday_playlist.collect():
    print(song[0], "-", song[1])

print("\nTotal songs:", birthday_playlist.count())
```
